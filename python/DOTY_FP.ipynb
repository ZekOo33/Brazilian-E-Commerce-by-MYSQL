```
import pandas as pd
import numpy as np
from datetime import datetime
```

```
# ── Load CSVs ──────────────────────────────────────────────────────────────────
df_orders_items = pd.read_csv(
    "/content/drive/MyDrive/dotpy_fp/orders_items_export.csv",
    on_bad_lines='skip',
    engine='python'
)
df_reviews      = pd.read_csv("/content/drive/MyDrive/dotpy_fp/reviews_export.csv")
df_geo          = pd.read_csv("/content/drive/MyDrive/dotpy_fp/geolocation_export.csv")

datasets = {
    "df_orders_items": df_orders_items,
    "df_reviews"     : df_reviews,
    "df_geo"         : df_geo,
}
```

```
# ── Function A — clean_location_text(x) ───────────────────────────────────────
def clean_location_text(x):
    """
    Cleans city or state text.
    - Missing values      → None
    - Length == 2         → UPPERCASE  (state code)
    - Otherwise           → Title Case (city name)
    - Empty after strip   → None
    """
    if x is None or (isinstance(x, float) and np.isnan(x)):
        return None

    x = str(x).strip()

    if not x:
        return None

    if len(x) == 2:
        return x.upper()

    return x.title()


# ── Function B — calc_delivery_days(purchase_dt, delivered_dt) ────────────────
def calc_delivery_days(purchase_dt, delivered_dt):
    """
    Calculates number of days between purchase and delivery.
    - Either date missing        → NaN
    - delivered < purchased      → NaN
    - Otherwise                  → int days difference
    """
    if pd.isnull(purchase_dt) or pd.isnull(delivered_dt):
        return np.nan

    purchase_dt  = pd.to_datetime(purchase_dt,  errors="coerce")
    delivered_dt = pd.to_datetime(delivered_dt, errors="coerce")

    if pd.isnull(purchase_dt) or pd.isnull(delivered_dt):
        return np.nan

    diff = (delivered_dt - purchase_dt).days

    if diff < 0:
        return np.nan

    return diff


# ── Function C — classify_review(score) ───────────────────────────────────────
def classify_review(score):
    """
    Maps a numeric review score to a sentiment label.
    - Missing  → "Unknown"
    - >= 4     → "Positive"
    - == 3     → "Neutral"
    - <= 2     → "Negative"
    """
    if pd.isnull(score):
        return "Unknown"

    score = int(score)

    if score >= 4:
        return "Positive"
    elif score == 3:
        return "Neutral"
    else:                   # score <= 2
        return "Negative"


# ── Function D — segment_customer(total_spend, total_orders, recency_days) ────
def segment_customer(total_spend, total_orders, recency_days):
    """
    Assigns a customer segment based on spend, order count, and recency.
    Priority order (first match wins):
      1. VIP      → spend >= 5000  OR  orders >= 10
      2. Loyal    → 3 <= orders <= 9
      3. New      → orders == 1  AND  recency <= 60 days
      4. At Risk  → recency > 180 days
      5. Regular  → everything else
    """
    if total_spend >= 5000 or total_orders >= 10:
        return "VIP"

    if 3 <= total_orders <= 9:
        return "Loyal"

    if total_orders == 1 and recency_days <= 60:
        return "New"

    if recency_days > 180:
        return "At Risk"

    return "Regular"
def classify_delivery_risk(avg_delivery_days):
    """
    Classifies delivery risk based on average delivery days.
    - Missing   → "Unknown"
    - < 5       → "Low Delay"
    - 5 – 10    → "Medium Delay"
    - > 10      → "High Delay"
    """
    if pd.isnull(avg_delivery_days):
        return "Unknown"

    if avg_delivery_days < 5:
        return "Low Delay"

    if avg_delivery_days <= 10:
        return "Medium Delay"

    return "High Delay"
```

```
for name, df in datasets.items():
    df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")
for name, df in datasets.items():
    print(f"   {name}: {df.columns.tolist()}")
```

```
# ══════════════════════════════════════════════════════════════════════════════
# STEP 4 — Fix Data Types
# ══════════════════════════════════════════════════════════════════════════════

# Numeric
df_orders_items["price"] = pd.to_numeric(df_orders_items["price"],         errors="coerce")
df_orders_items["freight_value"] = pd.to_numeric(df_orders_items["freight_value"], errors="coerce")
df_reviews["review_score"]       = pd.to_numeric(df_reviews["review_score"],       errors="coerce")

# Datetime
df_orders_items["order_purchase_timestamp"]      = pd.to_datetime(df_orders_items["order_purchase_timestamp"],      errors="coerce")
df_orders_items["order_delivered_customer_date"] = pd.to_datetime(df_orders_items["order_delivered_customer_date"], errors="coerce")
df_orders_items["order_estimated_delivery_date"] = pd.to_datetime(df_orders_items["order_estimated_delivery_date"], errors="coerce")

print("\n   df_orders_items dtypes:")
print(df_orders_items[["price", "freight_value",
                        "order_purchase_timestamp",
                        "order_delivered_customer_date"]].dtypes.to_string())
print("\n   df_reviews dtypes:")
print(df_reviews[["review_score"]].dtypes.to_string())
```

```
# ══════════════════════════════════════════════════════════════════════════════
# STEP 5 — Clean Location Fields
# ══════════════════════════════════════════════════════════════════════════════

df_orders_items["customer_city"]  = df_orders_items["customer_city"].apply(clean_location_text)
df_orders_items["customer_state"] = df_orders_items["customer_state"].apply(clean_location_text)

print(df_orders_items[["customer_city", "customer_state"]].drop_duplicates().head(5))


# ══════════════════════════════════════════════════════════════════════════════
# STEP 6 — Create Order-Level Business Columns
# ══════════════════════════════════════════════════════════════════════════════

# 1. order_month (YYYY-MM)
df_orders_items["order_month"] = df_orders_items["order_purchase_timestamp"].dt.to_period("M").astype(str)

# 2. order_year
df_orders_items["order_year"] = df_orders_items["order_purchase_timestamp"].dt.year

# 3. delivery_days — calls calc_delivery_days() via apply
df_orders_items["delivery_days"] = df_orders_items.apply(
    lambda row: calc_delivery_days(
        row["order_purchase_timestamp"],
        row["order_delivered_customer_date"]
    ),
    axis=1
)

# 4. delivery_status
df_orders_items["delivery_status"] = df_orders_items["order_delivered_customer_date"].apply(
    lambda x: "Delivered" if pd.notnull(x) else "Not Delivered"
)

# 5. late_delivery_flag
df_orders_items["late_delivery_flag"] = (
    df_orders_items["order_delivered_customer_date"] >
    df_orders_items["order_estimated_delivery_date"]
).astype(int)

print(df_orders_items[[
    "order_id", "order_month", "order_year",
    "delivery_days", "delivery_status", "late_delivery_flag"
]].head(5).to_string(index=False))


# ══════════════════════════════════════════════════════════════════════════════
# STEP 7 — Build Order Revenue Summary
# ══════════════════════════════════════════════════════════════════════════════

df_order_revenue = (
    df_orders_items.groupby("order_id")
    .agg(
        customer_id         = ("customer_id",    "first"),
        order_month         = ("order_month",    "first"),
        total_items_value   = ("price",          "sum"),
        total_freight_value = ("freight_value",  "sum"),
        items_count         = ("order_item_id",  "count"),
    )
    .reset_index()
)

df_order_revenue["total_order_value"] = (
    df_order_revenue["total_items_value"] +
    df_order_revenue["total_freight_value"]
)

df_order_revenue = df_order_revenue.sort_values("total_order_value", ascending=False).reset_index(drop=True)

print(f"   Shape: {df_order_revenue.shape[0]:,} rows × {df_order_revenue.shape[1]} columns")
print(df_order_revenue[[
    "order_id", "customer_id", "order_month",
    "total_items_value", "total_freight_value",
    "total_order_value", "items_count"
]].head(5).to_string(index=False))
```

```
# ══════════════════════════════════════════════════════════════════════════════
# STEP 8 — Merge Reviews with Orders
# ══════════════════════════════════════════════════════════════════════════════

df_orders_with_reviews = pd.merge(
    df_order_revenue,
    df_reviews[["order_id", "review_score", "review_creation_date"]],
    on        = "order_id",
    how       = "left"       # keep all orders, even those without a review
)

# review_level — calls classify_review()
df_orders_with_reviews["review_level"] = df_orders_with_reviews["review_score"].apply(
    classify_review
)

print("df_orders_with_reviews built.")
print(f"   Shape : {df_orders_with_reviews.shape[0]:,} rows × {df_orders_with_reviews.shape[1]} columns")
print(f"\n   Review level distribution:")
print(df_orders_with_reviews["review_level"].value_counts().to_string())
print()
print(df_orders_with_reviews[[
    "order_id", "total_order_value", "review_score", "review_level"
]].head(8).to_string(index=False))
# ══════════════════════════════════════════════════════════════════════════════
# STEP 9 — Customer Segmentation
# ══════════════════════════════════════════════════════════════════════════════

today = pd.Timestamp.today().normalize()

df_customer_summary = (
    df_order_revenue.groupby("customer_id")
    .agg(
        total_orders      = ("order_id",          "count"),
        total_spend       = ("total_order_value",  "sum"),
        avg_order_value   = ("total_order_value",  "mean"),
        last_purchase_date= ("order_month",        "max"),
    )
    .reset_index()
)

# recency_days — days since last purchase
df_customer_summary["last_purchase_date"] = pd.to_datetime(
    df_customer_summary["last_purchase_date"], format="%Y-%m", errors="coerce"
)
df_customer_summary["recency_days"] = (
    today - df_customer_summary["last_purchase_date"]
).dt.days

# customer_segment — FOR LOOP (mandatory)
segments = []
for _, row in df_customer_summary.iterrows():
    seg = segment_customer(
        total_spend   = row["total_spend"],
        total_orders  = row["total_orders"],
        recency_days  = row["recency_days"]
    )
    segments.append(seg)

df_customer_summary["customer_segment"] = segments

print("df_customer_summary built.")
print(f"   Shape : {df_customer_summary.shape[0]:,} rows × {df_customer_summary.shape[1]} columns")
print(f"\n   Segment distribution:")
print(df_customer_summary["customer_segment"].value_counts().to_string())
print(df_customer_summary.head(5).to_string(index=False))


# ══════════════════════════════════════════════════════════════════════════════
# STEP 10 — Product Performance Report
# ══════════════════════════════════════════════════════════════════════════════

df_product_summary = (
    df_orders_items.groupby("product_id")
    .agg(
        total_revenue    = ("price",       "sum"),
        total_orders     = ("order_id",    "nunique"),
        unique_customers = ("customer_id", "nunique"),
    )
    .reset_index()
)

# revenue_contribution_%
total_revenue = df_product_summary["total_revenue"].sum()
df_product_summary["revenue_contribution_%"] = (
    df_product_summary["total_revenue"] / total_revenue * 100
).round(4)

# product_class using numpy quantile
p50 = np.percentile(df_product_summary["total_revenue"], 50)   # bottom 50% cutoff
p80 = np.percentile(df_product_summary["total_revenue"], 80)   # top 20% cutoff

def classify_product(revenue):
    if revenue >= p80:
        return "Top"
    elif revenue >= p50:
        return "Mid"
    else:
        return "Low"

df_product_summary["product_class"] = df_product_summary["total_revenue"].apply(classify_product)

df_product_summary = df_product_summary.sort_values("total_revenue", ascending=False).reset_index(drop=True)

print("\n df_product_summary built.")
print(f"   Shape : {df_product_summary.shape[0]:,} rows × {df_product_summary.shape[1]} columns")
print(f"\n   Product class distribution:")
print(df_product_summary["product_class"].value_counts().to_string())
print(df_product_summary.head(5).to_string(index=False))


# ══════════════════════════════════════════════════════════════════════════════
# STEP 11 — Monthly Delivery Analysis
# ══════════════════════════════════════════════════════════════════════════════

df_month_delivery = (
    df_orders_items.groupby("order_month")
    .agg(
        avg_delivery_days = ("delivery_days",  "mean"),
        avg_freight_value = ("freight_value",  "mean"),
        total_orders      = ("order_id",       "nunique"),
    )
    .reset_index()
)

df_month_delivery["avg_delivery_days"] = df_month_delivery["avg_delivery_days"].round(2)
df_month_delivery["avg_freight_value"] = df_month_delivery["avg_freight_value"].round(2)

# delivery_risk — calls classify_delivery_risk()
df_month_delivery["delivery_risk"] = df_month_delivery["avg_delivery_days"].apply(
    classify_delivery_risk
)

df_month_delivery = df_month_delivery.sort_values("order_month").reset_index(drop=True)

print("Step 11 — df_month_delivery built.")
print(f"   Shape : {df_month_delivery.shape[0]:,} rows × {df_month_delivery.shape[1]} columns")
print(f"\n   Delivery risk distribution:")
print(df_month_delivery["delivery_risk"].value_counts().to_string())
print(df_month_delivery.head(8).to_string(index=False))
# ══════════════════════════════════════════════════════════════════════════════
# STEP 12 — Location Performance
# ══════════════════════════════════════════════════════════════════════════════
df_location_summary = (
    df_orders_items.groupby(["customer_state", "customer_city"])
    .agg(
        total_revenue    = ("price",          "sum"),
        unique_customers = ("customer_id",    "nunique"),
    )
    .reset_index()
)

df_location_summary["revenue_per_customer"] = (
    df_location_summary["total_revenue"] /
    df_location_summary["unique_customers"]
).round(2)

df_location_summary = df_location_summary.sort_values(
    "total_revenue", ascending=False
).reset_index(drop=True)

print("Step 12 — df_location_summary built.")
print(f"   Shape : {df_location_summary.shape[0]:,} rows × {df_location_summary.shape[1]} columns")
print(df_location_summary.head(10).to_string(index=False))
```

```
# ══════════════════════════════════════════════════════════════════════════════
# STEP 13 — Export Final CSV Files
# ══════════════════════════════════════════════════════════════════════════════

exports = {
    "final_monthly_sales.csv"       : df_month_delivery,
    "final_customer_segments.csv"   : df_customer_summary,
    "final_product_summary.csv"     : df_product_summary,
    "final_delivery_summary.csv"    : df_month_delivery,
    "final_location_summary.csv"    : df_location_summary,
    "final_orders_with_reviews.csv" : df_orders_with_reviews,
}

print("⏳ Exporting files...\n")

for filename, df in exports.items():
    df.to_csv(filename, index=False)
    print(f"    {filename:<40} → {df.shape[0]:,} rows × {df.shape[1]} columns")

print("\n All 6 files exported successfully.")
```
